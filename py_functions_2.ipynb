```python
# Write a function named find_largest() that accepts three numbers.

# Requirements
# Do not use the built-in max() function.
# Compare the numbers using conditional statements.
# Return the largest number.
# Accept three values from the user.
a = float(input("enter values of a"))
b = float(input("enter values of b"))
c = float(input("enter values of c"))
def find_largest(a,b,c):
    if a>b and a>c:
        return a,'a'
    elif b>c:
        return b,'b'
    else:
        return c,'c'
large,varible = find_largest(a,b,c)
print(large,varible)
```

Assignment 42: Sort Dictionaries by Age

```python
# Given a list of dictionaries containing names and ages, sort the list by age.
students = [
    {"name": "Ravi", "age": 24},
    {"name": "Anu", "age": 21},
    {"name": "Kiran", "age": 27}
]
youngest = students.copy()
youngest.sort(key=lambda students: students['age'])

oldest = students.copy()
oldest.sort(key=lambda students: students['age'],reverse = True)
print(youngest)
print(oldest)
```

Assignment 51: Add Two Lists Using map()

```python
# Given two lists of equal length, add corresponding values.

# Requirements
# Use map() with two iterables.
# Do not use an index-based loop.
# Validate that both lists contain the same number of elements.
inpit1 = input("enter number spearted by spaces ")
inpit2 = input("enter number spearted by spaces ")
list1 = list(map(int,inpit1.split()))
list2 = list(map(int,inpit2.split()))
print(list1,list2)
if len(list1) != len(list2):
    print('error')
else:
    result = list(map(lambda a,b : a+b,list1,list2))
    print(result)
```
